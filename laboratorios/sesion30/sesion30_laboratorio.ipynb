### Criterios de aceptación

- [ ] Un subárbol constante se sustituye por su valor y no genera instrucciones.

````python
%%writefile api_sesion30.h
/* api_sesion30.h - contrato de la Sesion 30
 * AST a TAC optimizado de punta a punta
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION30_H
#define API_SESION30_H

typedef struct Nodo {
    char tipo;
    char op;
    char valor[16];
    struct Nodo *izq;
    struct Nodo *der;
} Nodo;

int generar_tac_opt(const Nodo *raiz, char lineas[][64], int max, char *resultado, int cap);

#endif /* API_SESION30_H */
````

````python
%%writefile artesano_sesion30_c.c
/* artesano_sesion30_c.c - Fase 1: El Artesano (sin IA)
 * AST a TAC optimizado de punta a punta.
 *
 * Implementa lo que declara api_sesion30.h.
 * El harness compila este archivo junto con el suyo: no escribas main().
 */
#include "api_sesion30.h"
#include <ctype.h>
#include <stdio.h>
#include <stdlib.h>
#include <string.h>

typedef struct EstadoTac {
    char (*lineas)[64];
    int max;
    int count;
    int temps;
} EstadoTac;

static int es_entero_str(const char *s, long *valor_out) {
    char *fin = NULL;
    long val;

    if (s == NULL || *s == '\0' || isspace((unsigned char)*s)) {
        return 0;
    }

    val = strtol(s, &fin, 10);
    if (fin == s || *fin != '\0') {
        return 0;
    }

    if (valor_out != NULL) {
        *valor_out = val;
    }
    return 1;
}

static void copiar_lugar(char *dst, int cap, const char *src) {
    if (dst == NULL || cap <= 0 || src == NULL) {
        return;
    }
    snprintf(dst, (size_t)cap, "%s", src);
}

static void emitir_instruccion(EstadoTac *st, const char *texto) {
    if (st == NULL || st->lineas == NULL || st->count < 0 || st->count >= st->max) {
        return;
    }
    snprintf(st->lineas[st->count], 64, "%s", texto);
    st->count++;
}

static void evaluar_nodo(const Nodo *nodo, EstadoTac *st, char *lugar, int cap_lugar) {
    char izq_lugar[16] = {0};
    char der_lugar[16] = {0};
    long izq_const = 0;
    long der_const = 0;
    int izq_es_const;
    int der_es_const;
    char buf[64];
    int t_idx;

    if (nodo == NULL || lugar == NULL || cap_lugar <= 0) {
        return;
    }

    if (nodo->tipo == 'N' || nodo->tipo == 'V') {
        copiar_lugar(lugar, cap_lugar, nodo->valor);
        return;
    }

    if (nodo->tipo == 'B') {
        evaluar_nodo(nodo->izq, st, izq_lugar, (int)sizeof(izq_lugar));
        evaluar_nodo(nodo->der, st, der_lugar, (int)sizeof(der_lugar));

        izq_es_const = es_entero_str(izq_lugar, &izq_const);
        der_es_const = es_entero_str(der_lugar, &der_const);

        /* Plegado de constantes cuando ambos operandos son enteros literales */
        if (izq_es_const && der_es_const) {
            if (nodo->op == '+') {
                snprintf(lugar, (size_t)cap_lugar, "%ld", izq_const + der_const);
                return;
            }
            if (nodo->op == '-') {
                snprintf(lugar, (size_t)cap_lugar, "%ld", izq_const - der_const);
                return;
            }
            if (nodo->op == '*') {
                snprintf(lugar, (size_t)cap_lugar, "%ld", izq_const * der_const);
                return;
            }
            if (nodo->op == '/' && der_const != 0) {
                snprintf(lugar, (size_t)cap_lugar, "%ld", izq_const / der_const);
                return;
            }
        }

        /* Identidades algebraicas */
        if (nodo->op == '+') {
            if (der_es_const && der_const == 0) {
                copiar_lugar(lugar, cap_lugar, izq_lugar);
                return;
            }
            if (izq_es_const && izq_const == 0) {
                copiar_lugar(lugar, cap_lugar, der_lugar);
                return;
            }
        } else if (nodo->op == '*') {
            if ((izq_es_const && izq_const == 0) || (der_es_const && der_const == 0)) {
                copiar_lugar(lugar, cap_lugar, "0");
                return;
            }
            if (der_es_const && der_const == 1) {
                copiar_lugar(lugar, cap_lugar, izq_lugar);
                return;
            }
            if (izq_es_const && izq_const == 1) {
                copiar_lugar(lugar, cap_lugar, der_lugar);
                return;
            }
        }

        /* Emisión de instrucción TAC si no se plegó ni simplificó */
        t_idx = st->temps++;
        snprintf(lugar, (size_t)cap_lugar, "_t%d", t_idx);
        snprintf(buf, sizeof(buf), "%s = %s %c %s", lugar, izq_lugar, nodo->op, der_lugar);
        emitir_instruccion(st, buf);
    }
}

int generar_tac_opt(const Nodo *raiz, char lineas[][64], int max,
                    char *resultado, int cap) {
    EstadoTac st;

    if (resultado != NULL && cap > 0) {
        resultado[0] = '\0';
    }
    if (raiz == NULL || lineas == NULL || max <= 0 || resultado == NULL || cap <= 0) {
        return 0;
    }

    st.lineas = lineas;
    st.max = max;
    st.count = 0;
    st.temps = 0;

    evaluar_nodo(raiz, &st, resultado, cap);
    return st.count;
}
````

## Fase 2 · El Científico (10 min)

Redacta el contrato **C-R-E-O** (Contexto, Rol, Especificación, Output) para pedirle
la misma función a la IA, y pega el resultado en la celda de `ia_sesion30_c.c`. La
firma debe ser idéntica a la del contrato; si la IA la cambia, corrígela antes de
compilar.


````python
%%writefile ia_sesion30_c.c
/* ia_sesion30_c.c - Fase 2: El Cientifico
 *
 * Implementacion generada con la MISMA firma que declara api_sesion30.h.
 * No puede traer main().
 */
#include "api_sesion30.h"
#include <ctype.h>
#include <stdio.h>
#include <stdlib.h>

typedef struct ContextoOpt {
    char (*lineas)[64];
    int max;
    int emitidas;
    int temp_id;
} ContextoOpt;

static int parse_entero_lugar(const char *s, long *out_val) {
    char *end = NULL;
    long v;

    if (!s || *s == '\0' || isspace((unsigned char)*s)) {
        return 0;
    }

    v = strtol(s, &end, 10);
    if (end == s || *end != '\0') {
        return 0;
    }

    if (out_val) {
        *out_val = v;
    }
    return 1;
}

static void set_lugar(char *dst, int cap, const char *val) {
    if (!dst || cap <= 0 || !val) {
        return;
    }
    snprintf(dst, (size_t)cap, "%s", val);
}

static void push_tac(ContextoOpt *ctx, const char *instr) {
    if (!ctx || !ctx->lineas || ctx->emitidas < 0 || ctx->emitidas >= ctx->max) {
        return;
    }
    snprintf(ctx->lineas[ctx->emitidas], 64, "%s", instr);
    ctx->emitidas++;
}

static void bajar_expresion(const Nodo *nodo, ContextoOpt *ctx,
                            char *lugar_out, int cap_out) {
    char l_izq[16] = {0};
    char l_der[16] = {0};
    long v_izq = 0;
    long v_der = 0;
    int es_num_izq;
    int es_num_der;
    char temp_nombre[12];
    char linea_tac[64];

    if (!nodo || !lugar_out || cap_out <= 0) {
        return;
    }

    if (nodo->tipo == 'N' || nodo->tipo == 'V') {
        set_lugar(lugar_out, cap_out, nodo->valor);
        return;
    }

    if (nodo->tipo == 'B') {
        bajar_expresion(nodo->izq, ctx, l_izq, (int)sizeof(l_izq));
        bajar_expresion(nodo->der, ctx, l_der, (int)sizeof(l_der));

        es_num_izq = parse_entero_lugar(l_izq, &v_izq);
        es_num_der = parse_entero_lugar(l_der, &v_der);

        if (es_num_izq && es_num_der) {
            if (nodo->op == '+') {
                snprintf(lugar_out, (size_t)cap_out, "%ld", v_izq + v_der);
                return;
            }
            if (nodo->op == '-') {
                snprintf(lugar_out, (size_t)cap_out, "%ld", v_izq - v_der);
                return;
            }
            if (nodo->op == '*') {
                snprintf(lugar_out, (size_t)cap_out, "%ld", v_izq * v_der);
                return;
            }
            if (nodo->op == '/' && v_der != 0) {
                snprintf(lugar_out, (size_t)cap_out, "%ld", v_izq / v_der);
                return;
            }
        }

        if (nodo->op == '+') {
            if (es_num_der && v_der == 0) {
                set_lugar(lugar_out, cap_out, l_izq);
                return;
            }
            if (es_num_izq && v_izq == 0) {
                set_lugar(lugar_out, cap_out, l_der);
                return;
            }
        } else if (nodo->op == '*') {
            if ((es_num_izq && v_izq == 0) || (es_num_der && v_der == 0)) {
                set_lugar(lugar_out, cap_out, "0");
                return;
            }
            if (es_num_der && v_der == 1) {
                set_lugar(lugar_out, cap_out, l_izq);
                return;
            }
            if (es_num_izq && v_izq == 1) {
                set_lugar(lugar_out, cap_out, l_der);
                return;
            }
        }

        snprintf(temp_nombre, sizeof(temp_nombre), "_t%d", ctx->temp_id++);
        snprintf(linea_tac, sizeof(linea_tac), "%s = %s %c %s",
                 temp_nombre, l_izq, nodo->op, l_der);
        push_tac(ctx, linea_tac);
        set_lugar(lugar_out, cap_out, temp_nombre);
    }
}

int generar_tac_opt(const Nodo *raiz, char lineas[][64], int max,
                    char *resultado, int cap) {
    ContextoOpt ctx;

    if (resultado && cap > 0) {
        resultado[0] = '\0';
    }
    if (!raiz || !lineas || max <= 0 || !resultado || cap <= 0) {
        return 0;
    }

    ctx.lineas = lineas;
    ctx.max = max;
    ctx.emitidas = 0;
    ctx.temp_id = 0;

    bajar_expresion(raiz, &ctx, resultado, cap);
    return ctx.emitidas;
}
````
